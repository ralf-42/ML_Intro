# 1 | Understand

# 2 | Prepare

---

Die Daten, die dem vortrainierten Modell zugeführt werden, müssen möglicherweise in ein Format umgewandelt werden, das das Modell versteht. Dies kann beinhalten, die Daten zu normieren, fehlende Werte zu behandeln oder Kategorien zu kodieren.

# 3 | Modeling
---



An dieser Stelle wird das vortrainierte Modell importiert. Man könnte auch überlegen, ob das Modell weiter auf spezifischen Daten (Fine-Tuning) trainiert werden sollte.

**Wann vortrainierte Modelle (Transfer Learning)?**

**In der Praxis relevant wenn:**
- Keine eigenen Trainingsdaten in ausreichender Menge vorhanden sind — vortrainierte Modelle wie YOLO wurden auf Millionen von Bildern trainiert
- Standardaufgaben gelöst werden sollen (Objekterkennung, Klassifikation), für die robuste Pre-Trained Models verfügbar sind
- Schnelle Prototypen gebraucht werden: vortrainierte Modelle liefern ohne eigenes Training sofort Ergebnisse

**Nicht geeignet wenn:**
- Sehr spezifische Domänen vorliegen (z.B. medizinische Bilder, Satellitenaufnahmen), für die keine passenden Pre-Trained Models existieren → dann Fine-Tuning oder eigenes Training
- Die gelernten Konzepte stark von den Trainingsdaten abweichen → Pre-Trained Models übertragen nicht alle Domänen gleich gut

**Typischer Fehler:**
Das vortrainierte Modell direkt für eine Spezialdomäne nutzen ohne Fine-Tuning — dann sind die Ergebnisse oft enttäuschend. Für domänenspezifische Aufgaben immer zumindest einige Epochen Fine-Tuning mit eigenen Daten durchführen.
